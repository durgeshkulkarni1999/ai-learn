```python
from langchain_openai import ChatOpenAI
```

```python
llm = ChatOpenAI(model="gpt-4o")
```

```python
prompts = [
    {"role":"system", "content":"You are a javascript developer"},
    {"role":"user", "content":"difference between map and filter?"}
]
res = llm.invoke(prompts)
print(res.content)
```

```python
prompts = [
    {"role":"system", "content":"You are a python developer"},
    {"role":"user", "content":"difference between map and filter?"}
]
res = llm.invoke(prompts)
print(res.content)
```

### Prompt Templates - Dynamic Prompts

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
out = StrOutputParser()
```

```python
prompts = ChatPromptTemplate.from_messages([
    {"role":"system", "content":"You are a translator and translate inout in {language}"},
    {"role":"user", "content":"{query}"}
])
#prompts = prompts.format(language="Hindi", query="I am a developer")
```

### Chains - Runnable

```python
chains = prompts | llm | out | transform_case
chains
```

```python
res = chains.invoke({"language": "Hinglish", "query": "I am a developer"})
print(res)
```

```python
def transform_case(text):
    return text.upper()
```
