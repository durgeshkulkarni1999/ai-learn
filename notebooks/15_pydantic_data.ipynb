```python
from pydantic import BaseModel, Field
from typing import List

class User(BaseModel):
    name: str
    email: str
    age: int
```

```python
user = User(name="Durgesh", email="durgesh@example.com", age=23)

print(user)
```

```python
class Product(BaseModel):
    title: str = Field(max_length=100)
    description: str = Field(min_length=20)
    price: int = Field(gt=0,le=100)
```

```python
product = Product(title="Sample Product", description="This is a sample product description.", price=50)

print(product)
```

```python
product2 = Product(title="Sample Product2", description="This is a sample product description.2", price=50)
```

```python
class Order(BaseModel):
    id: str = Field(min_length=6,max_length=12)
    product: Product
```

```python
order = Order(id="123jud", product=product)
```

```python
print(order)
```

```python
print(order.product.title)
```

```python
class Carts(BaseModel):
    id: str = Field(min_length=6,max_length=12)
    items: list[Product]
    comment: str = Field(default="No comment")
```

```python
cart = Carts(id="123abc", items=[product,product2])
```

```python
print(cart)
```
