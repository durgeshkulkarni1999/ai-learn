```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_community.utilities import GoogleSerperAPIWrapper
from langchain_groq import ChatGroq
from langchain.agents import create_agent
from langgraph.checkpoint.memory import MemorySaver
```

```python
llm = ChatGroq(model="openai/gpt-oss-20b")
search = GoogleSerperAPIWrapper()
```

```python
agent = create_agent(
    model = llm,
    tools = [search.run],
    system_prompt = "You are a helpful assistant that can perform Google searches and always use the search tool for any queries.",
    checkpointer = MemorySaver()
)
```

```python
question = "What are total number of employees it have"
```

```python
response = agent.invoke({"messages" : [{"role": "user", "content": question}]},
                        {"configurable": {"thread_id" : "asd123"}})
```
