```python
from dotenv import load_dotenv
load_dotenv()
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_openai import OpenAIEmbeddings, ChatOpenAI
from langchain_chroma import Chroma
from langchain_core.prompts import PromptTemplate
```

```python
loader = PyPDFLoader("../data/medical_report.pdf")
docs = loader.load()
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=200)
splitted_data = splitter.split_documents(docs)
```

```python
embeddings = OpenAIEmbeddings(model="text-embedding-3-large")
```

```python
vector_store = Chroma.from_documents(documents=splitted_data, embedding=embeddings)
```

```python
query = "Machine Learning and Data Science Content"
data = vector_store.similarity_search(query)
```

```python
context = ""
for doc in data:
    context += doc.page_content + "\n"
print (context)
```

```python
llm = ChatOpenAI(model="gpt-5")
```

```python
def get_context(query: str):
    data = vector_store.similarity_search(query)
    context = ""
    for doc in data:
        context += doc.page_content + "\n"
    return {"context": context, "question": query} 
```

```python
prompt = PromptTemplate.from_template("""
You are a helpful assistant that answers questions based on the context provided., if you don't know the answer, just say that you don't know, don't try to make up an answer.
context : {context}
Question : {question}
""")
```

```python
rag_chain = get_context | prompt | llm
```

```python
res = rag_chain.invoke("What is pateint's name")
print(res.content)
```
