```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_core.tools import tool
```

```python
@tool
def add_numbers(a: int, b: int) -> int:
    """
    Add two numbers together.
    Args:
        a (int): The first number.
        b (int): The second number.
    """
    return a + b

@tool
def multiply_numbers(a: int, b: int) -> int:
    """
    Multiply two numbers together.
    Args:
        a (int): The first number.
        b (int): The second number.
    """
    return a * b
```

```python
add_numbers.invoke({"a": 5, "b": 3})
```

```python
from langchain_openai import ChatOpenAI
from langchain.agents import create_agent
```

```python
llm = ChatOpenAI(model="gpt-4o-mini")
```

```python
agent = create_agent(
    model = llm,
    tools = [add_numbers, multiply_numbers],
    system_prompt = "You are a helpful assistant that can perform basic arithmetic operations and always use tool for calculations."
)
```

```python
response = agent.invoke({"messages" : [{"role": "user", "content": "What is 5 + 3?"}]})
```
